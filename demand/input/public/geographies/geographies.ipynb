# Format geographies

We format the geojson files to have a unified structure.

```python
# Imports and general variables

import geopandas as gpd
```

```python
# Read the geojson files

municipality_gdp = gpd.read_file('georef-sweden-kommun@public.geojson', encoding='utf-8')
county_gdp = gpd.read_file('georef-sweden-lan.geojson', encoding='utf-8')
```

```python
# Rename the municipality columns to have a unified structure
municipality_gdp.rename(columns={'kom_type': 'geo_type', 'kom_code': 'geo_id', 'kom_name': 'geo_name'}, inplace=True)

# Drop the kom_area_code column
municipality_gdp.drop(columns=['kom_area_code', 'lan_code', 'lan_name'], inplace=True)

# Set the geo_type column to "municipality"
municipality_gdp['geo_type'] = 'municipality'
```

```python
# Rename the county columns to have a unified structure
county_gdp.rename(columns={'lan_type': 'geo_type', 'lan_code': 'geo_id', 'lan_name': 'geo_name'}, inplace=True)

# Drop the lan_area_code column
county_gdp.drop(columns=['lan_area_code'], inplace=True)

# Set the geo_type column to "county"
county_gdp['geo_type'] = 'county'
```

```python
# Write files
municipality_gdp.to_file('sweden-municipalities.geojson', driver='GeoJSON')
county_gdp.to_file('sweden-counties.geojson', driver='GeoJSON')
```
