# Generate growth time series

This notebook generates a time series of growth for a set of given scenarios.

```python
# Imports and general variables

import pandas as pd 
import numpy as np

start_time = pd.Timestamp("2025-01-01 00:00:00")
end_time = pd.Timestamp("2045-12-31 23:00:00")
resolution = '1h'
type = 'exponential_growth_target'
scenarios = [
    {'scenario': 1, 'target': 2},
    {'scenario': 2, 'target': 2.33},
    {'scenario': 3, 'target': 2.4},
]

scenario = scenarios[1]
```

```python
# Generate the new timestamps for the extended demand
timestamps = pd.date_range(start=start_time, end=end_time, freq=resolution)

delta_intervals = len(timestamps)
interval_growth =  scenario['target'] ** (1 / delta_intervals) - 1

# Calculate the growth factors for all hours
intervals_elapsed = np.arange(len(timestamps))  # Elapsed hours as a 1D array
growth_factors = (1 + interval_growth) ** intervals_elapsed  # Shape: (len(new_timestamps),)
```
