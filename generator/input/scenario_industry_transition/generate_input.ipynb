# Input for the industry transition scenario

This scenario captures the industry transition to using more electricity as the energy source.

At present it projects an s-curve growth of electricity demand between 2025 and 2050.

```python
# 1. Import libraries and set basic variables

import sys
import numpy as np
import pandas as pd
from pathlib import Path

sys.path.append(str(Path(globals()['_dh'][0]).resolve().parent.parent.parent))

from generator.library.curves import generate_s_curve

from generator.paths import input_path
```

```python
# 2. Set the scenario parameters for this input

scenario_ids = range(3)
scenario_targets = [97/34,(97+187)/(2*34),187/34]
start_year = 2025
end_year = 2050
```

```python
# 3. Generate the s-curves for each scenario and write to parquet files

all_curves = []

for i in scenario_ids:
    df = generate_s_curve(
        curve_id=i,
        start_year=start_year,
        end_year=end_year,
        y0=1.0,
        y1=scenario_targets[i],
        resolution="1h",
    ).to_parquet(f"scenario_industry_transition,i={i},start_year={start_year},end_year={end_year}.parquet", index=False)
```
