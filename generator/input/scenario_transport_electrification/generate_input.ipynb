# Input for the transport electrification scenario

This scenario captures the increased transport electricity demand. It is a dummy model to be replaced by a proper first-principle model of transport electricity demand.

At present it projects an s-curve growth of electricity demand between 2025 and 2050.

```python
# 1. Import libraries and set basic variables

import sys
import numpy as np
import pandas as pd
from pathlib import Path

sys.path.append(str(Path(globals()['_dh'][0]).resolve().parent.parent.parent))

from generator.library.curves import generate_s_curve

from generator.paths import input_path
```

```python
# 2. Set the scenario parameters for this input

scenario_ids = range(5)
scenario_targets = [6,8,10,12,14]
start_year = 2025
end_year = 2050
```

```python
# 3. Generate the s-curves for each scenario and write to parquet files

all_curves = []

for i in scenario_ids:
    df = generate_s_curve(
        curve_id=i,
        start_year=start_year,
        end_year=end_year,
        y0=1.0,
        y1=scenario_targets[i],
        freq="1h",
    ).to_parquet(f"scenario_transport_electrification,i={i},start_year={start_year},end_year={end_year}.parquet", index=False)
```
